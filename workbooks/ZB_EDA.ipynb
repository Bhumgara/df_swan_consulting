```python
no_tenure_df = df.drop(columns=["Tenure Months"]).copy()
```

```python
def build_ydf_model(df, test_size:float = .2, random_state:int = 1204, num_trials:int = 50):
    features = list(df.columns)
    features.remove('Churn Value')

    y = df['Churn Value']
    X = df[features]
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state, stratify=y)

    X_train['Churn Value'] = y_train # This is adding the label/target back to X train
    X_test['Churn Value'] = y_test   # Same for X_test

    tuner = ydf.RandomSearchTuner(num_trials=num_trials)
    model = ydf.GradientBoostedTreesLearner(tuner=tuner, label='Churn Value').train(X_train)

    return model
```

```python
def model_metrics(model):
    train_eval = model.evaluate(X_train)
    test_eval = model.evaluate(X_test)
    plot = model.plot_tree(max_depth=3)
    desc = model.describe()
    importance = model.variable_importances()
    return train_eval, test_eval, plot, desc, importance
```

```python
model2 = build_ydf_model(no_tenure_df)
```

```python
train_eval, test_eval, plot, desc, importance = model_metrics(model2)
```

```python
long_tenure_df = df[df["Tenure Months"] >= 2].copy()

long_tenure_df.drop(columns=["Tenure Months"], inplace=True)

model2 = build_ydf_model(long_tenure_df)
train_eval, test_eval, plot, desc, importance = model_metrics(model2)
```
